```python
import os
import pandas as pd

# 1. Search recursively for the CSV files in Kaggle input
def find_file(filename, search_path='/kaggle/input'):
    for root, dirs, files in os.walk(search_path):
        if filename in files:
            return os.path.join(root, filename)
    return None

mql_path = find_file('olist_marketing_qualified_leads_dataset.csv')
deals_path = find_file('olist_closed_deals_dataset.csv')

# Verify files were located
if not mql_path or not deals_path:
    raise FileNotFoundError(
        "Could not locate one or both CSV files. "
        "Please make sure 'Marketing Funnel by Olist' is added under the Notebook's 'Input' section."
    )

# 2. Load the Olist Marketing Funnel files into Pandas DataFrames
mql_df = pd.read_csv(mql_path)
closed_deals_df = pd.read_csv(deals_path)

# 3. Calculate Core Funnel Statistics
total_leads = len(mql_df)
closed_wins = len(closed_deals_df)

# Calculate conversion rate from visitor lead to customer purchase
conversion_rate = round((closed_wins / total_leads) * 100, 2)

# 4. Marketing Channels Allocation Analysis
leads_by_channel = mql_df['origin'].value_counts().reset_index()
leads_by_channel.columns = ['Marketing_Channel', 'Leads_Generated']

# 5. Display the calculated growth metrics summary
print("Digital Marketing Campaign Optimization Framework Live!\n")
print(f"Total Funnel Leads Monitored: {total_leads}")
print(f"Total Successful Conversions: {closed_wins}")
print(f"Overall Lead-to-Customer Conversion Rate: {conversion_rate}%")
print("\n--- Top Lead Generation Channels ---")
print(leads_by_channel.head(5))
```
